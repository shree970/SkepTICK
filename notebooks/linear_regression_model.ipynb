```python
import statsmodels.api as sm
import pandas as pd
```

```python
df = pd.read_csv("updated_graph_extractor.csv")
```

```python
from sklearn.model_selection import train_test_split

# Define the dependent (response) variable and independent (predictor) variables
X = df[["Subscribers", "Num Videos", "Registered"]]
y = df["Credibility"]

# Split the data into train and test sets using stratified sampling based on the "Registered" column
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=X["Registered"], random_state=42)

# Check the proportions in train and test sets
train_proportion = X_train["Registered"].value_counts(normalize=True)
test_proportion = X_test["Registered"].value_counts(normalize=True)

train_proportion, test_proportion
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Initialize the linear regression model
lr = LinearRegression()

# Fit the model to the training data
lr.fit(X_train, y_train)

# Predict the credibility scores on the test set
y_pred = lr.predict(X_test)

# Evaluate the model using various metrics
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

mae, mse, r2
```

```python
# Define the dependent (response) variable and independent (predictor) variables
X = df[["Subscribers", "Num Videos", "Registered"]]
y = df["Credibility"]

# Add a constant to the predictor variables (for the intercept term)
X = sm.add_constant(X)

# Perform linear regression
model = sm.OLS(y, X).fit()

# Get the regression summary
summary = model.summary()
summary
```
